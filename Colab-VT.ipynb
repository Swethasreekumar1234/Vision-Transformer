```
import os
from pathlib import Path

# List contents of /content (Colab's root directory)
print("Contents of /content:")
for item in os.listdir('/content'):
    path = os.path.join('/content', item)
    if os.path.isdir(path):
        print(f"📁 {item}/")
    else:
        print(f"📄 {item}")

# Try to find your dataset
print("\n" + "="*60)
print("Searching for dataset folders...")
print("="*60)

for root, dirs, files in os.walk('/content'):
    # Look for common dataset structures
    if any(d in dirs for d in ['train', 'valid', 'test']):
        print(f"\n✅ Found dataset structure at:")
        print(f"   {root}")

        # Show what's in it
        for d in ['train', 'valid', 'test']:
            if d in dirs:
                subpath = os.path.join(root, d)
                files_count = len([f for f in os.listdir(subpath) if f.endswith(('.jpg', '.jpeg', '.png', '.txt'))])
                print(f"     ├── {d}/ ({files_count} files)")
```

```
import os
from pathlib import Path

# Check the actual structure
for split in ['train', 'valid', 'test']:
    split_path = f'/content/{split}'
    if os.path.exists(split_path):
        print(f"\n📁 {split}/")
        for item in os.listdir(split_path):
            item_path = os.path.join(split_path, item)
            if os.path.isdir(item_path):
                files = len(os.listdir(item_path))
                print(f"   ├── 📁 {item}/ ({files} files)")
            else:
                print(f"   ├── 📄 {item}")
```
